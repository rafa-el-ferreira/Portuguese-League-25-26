# **BONUS QUESTION: LEAGUE STANDINGS**

```python
#the goal is to count how many times a team won at home and away
#do the same with draw

home_wins = PPL_matches[PPL_matches["score.winner"] == "HOME_TEAM"]["homeTeam.name"].value_counts()
away_wins = PPL_matches[PPL_matches["score.winner"] == "AWAY_TEAM"]["awayTeam.name"].value_counts()

draw_home = PPL_matches[PPL_matches["score.winner"] == "DRAW"]["homeTeam.name"].value_counts()
draw_away = PPL_matches[PPL_matches["score.winner"] == "DRAW"]["awayTeam.name"].value_counts()
```

```python
#These are the functions with all the results:
#home_wins
#away_wins
#draw_home
#draw_away

#let's passe them to DataFrame
home_wins = home_wins.reset_index()
away_wins = away_wins.reset_index()
draw_home = draw_home.reset_index()
draw_away = draw_away.reset_index()
```

```python
# Rename columns to standardize them before merging
home_wins_df = home_wins.rename(columns={'homeTeam.name': 'Team', 'count': 'Home Wins'})
away_wins_df = away_wins.rename(columns={'awayTeam.name': 'Team', 'count': 'Away Wins'})
draw_home_df = draw_home.rename(columns={'homeTeam.name': 'Team', 'count': 'Home Draws'})
draw_away_df = draw_away.rename(columns={'awayTeam.name': 'Team', 'count': 'Away Draws'})

# Merge DataFrames using an outer merge to include all teams

league_standings = pd.merge(home_wins_df, away_wins_df, on='Team', how='outer')
league_standings = pd.merge(league_standings, draw_home_df, on='Team', how='outer')
league_standings = pd.merge(league_standings, draw_away_df, on='Team', how='outer')

# Fill NaN values (teams with no wins/draws in a category) with 0
league_standings = league_standings.fillna(0)

# Convert count columns to integers
league_standings['Home Wins'] = league_standings['Home Wins'].astype(int)
league_standings['Away Wins'] = league_standings['Away Wins'].astype(int)
league_standings['Home Draws'] = league_standings['Home Draws'].astype(int)
league_standings['Away Draws'] = league_standings['Away Draws'].astype(int)

# New DataFrame

league_standings
```

```python
# Calculate all the points to each team
league_standings['Points'] = (league_standings['Home Wins'] + league_standings['Away Wins']) * 3 + \
                             (league_standings['Home Draws'] + league_standings['Away Draws']) * 1

# Order them by Points
league_standings_final = league_standings.sort_values(by='Points', ascending=False).reset_index(drop=True)

# Final DataFrame
league_standings_final[["Team", "Points"]]
```
