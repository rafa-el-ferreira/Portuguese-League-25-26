```python
#holding on wins at home (win at HT and FT)

first_metric = PPL_matches[(PPL_matches["score.halfTime.home"] > PPL_matches["score.halfTime.away"]) & (PPL_matches["score.winner"] == "HOME_TEAM")]
```

```python
first_metric
```

```python
# show what we really want
first_metric["homeTeam.name"].value_counts()
```

What we really want to see is the % of wins at HT that ended on a win at FT

```python
#check, first, the amount of times that teams were winning at HT

first_metric_final = PPL_matches[PPL_matches["score.halfTime.home"] > PPL_matches["score.halfTime.away"]]
first_metric_final["homeTeam.name"].value_counts()
```

```python
#Now see how many of those times ended up on winning at FT and put it on %

ab = (first_metric["homeTeam.name"].value_counts() / first_metric_final["homeTeam.name"].value_counts() * 100).round(1)
ab.sort_values(ascending=False)
```

# **2nd METRIC** **MAIN** **CODE**

```python
#turning the score around in the second half

#first is to see the turnarounds in the home games

second_metric_home = PPL_matches[(PPL_matches["score.halfTime.home"] < PPL_matches["score.halfTime.away"]) & (PPL_matches["score.winner"] == "HOME_TEAM")]
```

```python
#second is to see the turnarouds in the away games

second_metric_away = PPL_matches[(PPL_matches["score.halfTime.away"] < PPL_matches["score.halfTime.home"]) & (PPL_matches["score.winner"] == "HOME_AWAY")]
```

```python
#Let's count them
second_metric_home["homeTeam.name"].value_counts()
```

```python
second_metric_home2 = PPL_matches[PPL_matches["score.halfTime.home"] < PPL_matches["score.halfTime.away"]]
second_metric_home2["homeTeam.name"].value_counts()
```

```python
bc = (second_metric_home["homeTeam.name"].value_counts() / second_metric_home2["homeTeam.name"].value_counts() * 100).round(1)
bc.sort_values(ascending=False)
```
